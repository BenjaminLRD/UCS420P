```python
#3) i 
import random
random.seed(1024170084)
numbers = [random.randint(100, 900) for i in range(100)]
#ii
odd_numbers = [x for x in numbers if x%2!= 0]
print("Odd numbers:", odd_numbers)
print("no. of odd numbers:", len(odd_numbers))
#iii
even_numbers = [x for x in numbers if x%2== 0]
print("Even numbers:",even_numbers)
print("no. of even numbers:",len(even_numbers))
#iv
def is_prime(num):
    if num < 2:
        return False
    
    for i in range(2, num):
        if num % i == 0:
            return False
        else:
            return True
prime_numbers = [x for x in numbers if is_prime(x)]

print("Prime numbers:", prime_numbers)
print("no. of prime numbers:", len(prime_numbers))
#v
most = max(numbers, key=numbers.count)
print("Most frequent number: ",most,", Amount of appearances: ", numbers.count(most))
```

```python
##4) i
roll= [1, 0, 2, 4, 1, 7, 0, 0]
A= {x * 7 for x in roll}
B={x * 9 for x in roll}
print("A=", A)
print("B=", B)

#1
union= A.union(B)
print("AUB=", union)
#2
intersection = A.intersection(B)
print("A^B =", intersection)
#3
A_minus_B= A.difference(B)
B_minus_A= B.difference(A)
print("A-B=", A_minus_B)
print("B-A=", B_minus_A)
#4
A= {0, 7, 14, 28, 49}
B= {0, 9, 18, 36, 63}
sym_diff= A.symmetric_difference(B)
print("A(-)B =", sym_diff)
#5
A = {0, 7, 14, 28, 49}
B = {0, 9, 18, 36, 63}
print("A is subset of B:", A.issubset(B))
print("B is superset of A:", B.issuperset(A))
```
