```python
# Assignment 5.2: WAP to add all numbers divisible by 7 and 9 from 1 to n
n=int(input("Enter number: "))
sum=0

for i in range(1,n+1):
    if i % 7==0 and i % 9==0:
        sum+=i
print(sum)
```

```python
# Assignment 5.3: WAP to add all prime numbers from 1 to n
n=int(input("Enter number: "))
sum=0

for i in range(2,n+1):
    prime=True
    for j in range(2,i):
        if i % j==0:
            prime=False
            break
    if prime:
        sum+=i
print(sum)
```

```python
# Assignment 6.1: WAP using function that add all odd numbers from 1 to n
def odd_sum(n):
    sum=0
    for i in range(1,n+1):
        if i % 2!=0:
            sum+=i
    return sum
n=int(input("Enter number: "))
print(odd_sum(n))

```

```python
# Assignment 6.2: WAP using function that add all prime numbers from 1 to n
def prime_sum(n):
    sum=0
    for i in range(2,n+1):
        prime=True
        for j in range(2,i):
            if i % j==0:
                prime=False
                break
        if prime:
            sum+=i
    return sum

n=int(input("Enter number: "))
print(prime_sum(n))
```
