```python
def score_query(query,df): 
    query= query.lower()
    result= []

    for i, row in df.iterrows():
        score= sum(word in row["keywords"].split() for word in query.split())
        if score>0:
            result.append([row["question"], row["answer"],score])

    return pd.DataFrame(
    result,
    columns=["question", "answer", "score"]).sort_values("score", ascending= False)

score_query("fee payment",df)
```

```python
def same_category(category_name, df):
    return df[df["category"] == category_name]
same_category("general",df)
```
